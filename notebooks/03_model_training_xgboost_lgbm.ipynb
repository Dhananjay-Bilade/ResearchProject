```python
import pandas as pd
import numpy as np
import time
import joblib
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

# 1. Load Processed Datasets
print("Loading processed training and test data...")
X_train = pd.read_csv('../data/processed/X_train_smote.csv')
y_train = pd.read_csv('../data/processed/y_train_smote.csv').values.ravel()
X_test = pd.read_csv('../data/processed/X_test.csv')
y_test = pd.read_csv('../data/processed/y_test.csv').values.ravel()

# Calculate scale_pos_weight
ratio = float(np.sum(y_train == 0) / np.sum(y_train == 1))

# 2. Train XGBoost Model
print("\n--- Training XGBoost Model on PaySim Dataset ---")
start_time = time.time()
xgb_model = XGBClassifier(
    n_estimators=150,
    max_depth=6,
    learning_rate=0.08,
    scale_pos_weight=ratio,
    tree_method='hist', # Fast training on large datasets
    random_state=42,
    n_jobs=-1
)
xgb_model.fit(X_train, y_train)
xgb_train_time = time.time() - start_time
print(f"XGBoost Training Complete in {xgb_train_time:.2f} seconds.")

# 3. Train LightGBM Model
print("\n--- Training LightGBM Model on PaySim Dataset ---")
start_time = time.time()
lgb_model = LGBMClassifier(
    n_estimators=150,
    max_depth=6,
    learning_rate=0.08,
    scale_pos_weight=ratio,
    random_state=42,
    n_jobs=-1
)
lgb_model.fit(X_train, y_train)
lgb_train_time = time.time() - start_time
print(f"LightGBM Training Complete in {lgb_train_time:.2f} seconds.")

# 4. Save Models into 'models/' Directory
joblib.dump(xgb_model, '../models/xgboost_paysim.pkl')
joblib.dump(lgb_model, '../models/lightgbm_paysim.pkl')
print("\nModels successfully saved in 'models/' folder!")
```
